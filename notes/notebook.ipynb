```python
def dcf1(d1, r, g, n) :
    npv = 0
    for i in range(n) :
        npv += d1 * (1 + g) ** (1 + i) / (1 + r) ** (1 + i)
    return npv
```

```python
value = dcf1(1000, 0.05, 0.03, 10)
print(value)
```

```python
cashflow_A = [-120, 10, 30, 50, 40, 10]
cashflow_B = [-120, 30, 40, 40, 20, 10]
def dcf2(cashflow_A, r) :
    total = 0
    for i, cash in enumerate(cashflow_A) :
        total += cash / (1 + r) ** i
    return total
```

```python
#culculate irr
def irr (cashflow, iteration = 10000) :
    rate_lower = 0
    rate_upper = 1
    current_irr = (rate_lower + rate_upper) / 2
    npv = dcf2(cashflow, current_irr)
    while abs(npv) > 0.001 and iteration >= 0 : # set accuracy
        if npv > 0 :
            rate_lower = current_irr
        else :
            rate_upper = current_irr
        current_irr = (rate_lower + rate_upper) / 2   
        npv = dcf2(cashflow, current_irr)
        iteration -= 1
    return current_irr

irr_A = irr(cashflow_A)
irr_B = irr(cashflow_B)
print(irr_A)
print(irr_B)
```
