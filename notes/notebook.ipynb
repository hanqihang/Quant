```python
def dcf(d1, r, g, n) :
    npv = 0
    for i in range(n) :
        npv += d1 * (1 + g) ** (1 + i) / (1 + r) ** (1 + i)
    return npv
```

```python
value = dcf(1000, 0.05, 0.03, 10)
print(value)
```
